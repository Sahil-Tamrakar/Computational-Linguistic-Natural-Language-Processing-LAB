** Experiment-1: Data Analysis**

```python
# Import required libraries for data manipulation and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Prompt user to manually select and upload 'employee_information_100.csv' from local machine
uploaded = files.upload()

# Read the uploaded CSV file into a Pandas DataFrame
df = pd.read_csv('employee_information_100.csv')

# Verify successful load by printing confirmation and displaying the first 5 records
print("Dataset successfully uploaded and loaded into DataFrame!\n")
df.head()
```

Experiment 1.1: Average Salary of Each Department (Bar Chart)

```python
# Group data by 'Department' and calculate the mean 'Salary' for each department
avg_salary_dept = df.groupby('Department')['Salary'].mean().reset_index()

# Plot the average salary per department using Seaborn barplot
plt.figure(figsize=(8, 5))
sns.barplot(data=avg_salary_dept, x='Department', y='Salary', palette='Blues_d')

# Add titles and labels for visual clarity
plt.title('Experiment 1.1: Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.xticks(rotation=45)  # Rotate department names for better readability
plt.show()
```

Experiment 1.2: Number of Employees Working in Each Department

```python
# Count employees per department
dept_emp_count = df['Department'].value_counts().reset_index()
dept_emp_count.columns = ['Department', 'Number of Employees']

print("Experiment 1.2: Employee Count per Department:")
display(dept_emp_count)
```

Experiment 1.2: Number of Employees Working in Each Department

```python
# Count employees per department
dept_emp_count = df['Department'].value_counts().reset_index()
dept_emp_count.columns = ['Department', 'Number of Employees']

print("Experiment 1.2: Employee Count per Department:")
display(dept_emp_count)
```

Experiment 1.3: Percentage of Male and Female Employees (Pie Chart)

```python
# Count the total number of male and female employees
gender_dist = df['Gender'].value_counts()

# Create a Pie Chart to visualize gender distribution percentage
plt.figure(figsize=(6, 6))
plt.pie(gender_dist,
        labels=gender_dist.index,
        autopct='%1.1f%%',       # Display percentage values rounded to 1 decimal
        colors=['#66b3ff', '#ff9999'],
        startangle=90)

plt.title('Experiment 1.3: Gender Distribution (Percentage)')
plt.show()
```

Experiment 1.4: Salary Distribution (Histogram)

```python
# Plot a histogram to show how employee salaries are distributed across bins
plt.figure(figsize=(8, 5))
sns.histplot(df['Salary'], kde=True, bins=12, color='teal') # kde=True adds a smooth density curve

plt.title('Experiment 1.4: Salary Distribution Histogram')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```

Experiment 1.5: Relationship Between Experience and Salary (Scatter Plot)

```python
# Create a scatter plot mapping 'Experience' on X-axis and 'Salary' on Y-axis
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Experience_Years', y='Salary', hue='Department', s=80)

# Add grid and labels for proper scatter plot analysis
plt.title('Experiment 1.5: Experience vs. Salary Relationship')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary')
plt.grid(True)
plt.show()
```

Experiment 1.6: Top 10 Employees with Highest Salary

```python
# Use nlargest() to extract the 10 rows with the highest values in 'Salary' column
top_10_salaries = df.nlargest(10, 'Salary')

# Output the result
print("Experiment 1.6: Top 10 Employees with Highest Salary:")
display(top_10_salaries)
```

Experiment 1.7: Highest Salary in Every Department

```python
# Group data by 'Department' and aggregate using max() on the 'Salary' column
max_salary_dept = df.groupby('Department')['Salary'].max().reset_index()

print("Experiment 1.7: Maximum Salary in Each Department:")
display(max_salary_dept)
```

Experiment 1.8: Employees Earning Above Overall Average Salary

```python
# Step 1: Calculate overall mean salary of all employees
overall_avg_salary = df['Salary'].mean()

# Step 2: Filter rows where employee salary is strictly greater than overall average
above_avg_employees = df[df['Salary'] > overall_avg_salary]

# Step 3: Print summary statistics and display filtered employees
print(f"Overall Average Salary: ${overall_avg_salary:.2f}")
print(f"Experiment 1.8: Total Employees Earning Above Average = {len(above_avg_employees)}")
display(above_avg_employees)
```

Experiment 1.9: Average Years of Experience in Each Department

```python
# Group by department and compute mean experience
avg_exp_dept = df.groupby('Department')['Experience_Years'].mean().reset_index()

# Rename columns appropriately
avg_exp_dept.columns = ['Department', 'Average Experience (Years)']

print("Experiment 1.9: Average Experience per Department:")
display(avg_exp_dept)
```

Experiment 1.10: Distribution of Employee Ages (Histogram)

```python
# Draw a histogram to visualize age distribution among employees
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], kde=True, bins=10, color='darkorange')

plt.title('Experiment 1.10: Employee Age Distribution')
plt.xlabel('Age (Years)')
plt.ylabel('Frequency')
plt.show()
```
