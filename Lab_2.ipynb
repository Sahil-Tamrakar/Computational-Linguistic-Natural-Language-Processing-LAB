# **Experiment-2: Basic Text Preprocessing**

`Experiment 2.1:`  Write a Python program to perform basic text preprocessing on a given data. The program should:


1.   Convert text to lowercase. (Also display how many uppercase?)
1.   Remove punctuation.(Also display how many different types of punctuations?)
1.   Remove numbers.(Also display how many numbers?)
2.   Remove extra whitespace. (How much extra whitespace is present?)
2.   Display the cleaned text.

`Experiment 2.2:` Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.
           (Three separate code: using NLTK, using spaCy  and without using NLTK/spacy)

`Experiment 2.3:` Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).


```python
# Install libraries
!pip install -q nltk spacy python-docx
!python -m spacy download en_core_web_sm -q

import re, string, nltk, spacy
from google.colab import files

# Helper function to read text from .txt or .docx
def load_text_from_upload(uploaded_dict):
    filename = list(uploaded_dict.keys())[0]
    if filename.endswith('.docx'):
        import docx
        doc = docx.Document(filename)
        return "\n".join([p.text for p in doc.paragraphs if p.text.strip()])
    else:
        with open(filename, 'r', encoding='utf-8', errors='ignore') as f:
            return f.read()

# Download NLTK models
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords

nlp = spacy.load('en_core_web_sm')
print("Setup complete! Proceed to upload your files individually below.")
```

Experiment 2.1: Basic Text Preprocessing

```python
print("Upload file for Experiment 2.1 (text_data):")
uploaded_2_1 = files.upload()
text_data = load_text_from_upload(uploaded_2_1)

# 1. Lowercase & Uppercase count
uppercase_count = sum(1 for c in text_data if c.isupper())
lowercase_text = text_data.lower()

# 2. Punctuation removal & types count
found_punctuations = set(c for c in text_data if c in string.punctuation)
no_punct_text = lowercase_text.translate(str.maketrans('', '', string.punctuation))

# 3. Number removal & count
numbers_found = re.findall(r'\d+', text_data)
no_numbers_text = re.sub(r'\d+', '', no_punct_text)

# 4. Extra whitespace removal & count
extra_spaces = len(re.findall(r'[ \t]{2,}', no_numbers_text))
cleaned_text = re.sub(r'\s+', ' ', no_numbers_text).strip()

# Results
print("\n=== EXPERIMENT 2.1 RESULTS ===")
print(f"• Uppercase Letters Count  : {uppercase_count}")
print(f"• Punctuation Types Count  : {len(found_punctuations)} -> {found_punctuations}")
print(f"• Numbers Count            : {len(numbers_found)} -> {numbers_found}")
print(f"• Extra Spaces Count       : {extra_spaces}\n")
print("--- CLEANED TEXT ---")
print(cleaned_text)
```

Experiment 2.2: Tokenization (3 Methods)

```python
print("Upload file for Experiment 2.2 (tokenization_data):")
uploaded_2_2 = files.upload()
tokenization_data = load_text_from_upload(uploaded_2_2)

print("\n=== EXPERIMENT 2.2 RESULTS ===")

# 1. Without NLTK / spaCy (Python Built-ins)
print("\n--- 1. WITHOUT NLTK / SPACY ---")
py_sentences = [s.strip() for s in re.split(r'[.!?]+', tokenization_data) if s.strip()]
py_words = re.findall(r'\b\w+\b', tokenization_data)
print(f"Sentences ({len(py_sentences)}):", py_sentences)
print(f"Words ({len(py_words)}):", py_words)

# 2. Using NLTK
print("\n--- 2. USING NLTK ---")
nltk_sentences = sent_tokenize(tokenization_data)
nltk_words = word_tokenize(tokenization_data)
print(f"Sentences ({len(nltk_sentences)}):", nltk_sentences)
print(f"Words ({len(nltk_words)}):", nltk_words)

# 3. Using spaCy
print("\n--- 3. USING SPACY ---")
doc = nlp(tokenization_data)
spacy_sentences = [sent.text.strip() for sent in doc.sents]
spacy_words = [token.text for token in doc]
print(f"Sentences ({len(spacy_sentences)}):", spacy_sentences)
print(f"Words ({len(spacy_words)}):", spacy_words)
```

Experiment 2.3: Stop Words Removal

```python
print("Upload file for Experiment 2.3 (clean_data):")
uploaded_2_3 = files.upload()
clean_data = load_text_from_upload(uploaded_2_3)

# Tokenize text
tokens = word_tokenize(clean_data)
stop_words = set(stopwords.words('english'))

# Separate stop words and filtered tokens
found_stopwords = [t for t in tokens if t.lower() in stop_words]
filtered_tokens = [t for t in tokens if t.lower() not in stop_words and t.isalnum()]

print("\n=== EXPERIMENT 2.3 RESULTS ===")
print("• ORIGINAL TOKENS:")
print(tokens)

print("\n• STOP WORDS FOUND IN TEXT:")
print(list(set(found_stopwords)))

print("\n• FILTERED TOKENS (NO STOP WORDS):")
print(filtered_tokens)
```
