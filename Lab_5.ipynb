Experiment 5.1: BPE Subword Tokenization
5.1 (a) Using Pretrained BPE Tokenizer (GPT-2)

```python
from transformers import AutoTokenizer
import pandas as pd

# Load pretrained GPT-2 tokenizer (uses BPE)
bpe_tokenizer = AutoTokenizer.from_pretrained("gpt2")

# Tokenize input text
tokens_51a = bpe_tokenizer.tokenize(text_data_5)
token_ids_51a = bpe_tokenizer.convert_tokens_to_ids(tokens_51a)

# Display tokens and IDs
df_51a = pd.DataFrame({'Token': tokens_51a, 'Token ID': token_ids_51a})
print("=== 5.1 (a) BPE TOKENIZATION (PRETRAINED GPT-2) ===")
display(df_51a.head(20))  # Displaying first 20 tokens
```

5.1 (b) BPE Tokenizer Trained from Scratch (Without Pretrained Model)

```python
from tokenizers import Tokenizer, models, trainers, pre_tokenizers
import pandas as pd

# 1. Save text to temporary file for training
with open("temp_bpe_corpus.txt", "w", encoding="utf-8") as f:
    f.write(text_data_5)

# 2. Initialize a BPE Model from scratch
tokenizer_scratch = Tokenizer(models.BPE())
tokenizer_scratch.pre_tokenizer = pre_tokenizers.Whitespace()

# 3. Train BPE model on your custom dataset file
trainer = trainers.BpeTrainer(vocab_size=1000, special_tokens=["[PAD]", "[UNK]"])
tokenizer_scratch.train(files=["temp_bpe_corpus.txt"], trainer=trainer)

# 4. Tokenize text
encoding = tokenizer_scratch.encode(text_data_5)

# Display generated tokens and IDs
df_51b = pd.DataFrame({'Token': encoding.tokens, 'Token ID': encoding.ids})
print("=== 5.1 (b) BPE TOKENIZATION (TRAINED FROM SCRATCH) ===")
display(df_51b.head(20))
```

Experiment 5.2: SentencePiece Tokenization
5.2 (a) Using Pretrained SentencePiece Tokenizer (ALBERT / XLNet)

```python
from transformers import AutoTokenizer
import pandas as pd

# Load pretrained ALBERT tokenizer (uses SentencePiece)
sp_pretrained = AutoTokenizer.from_pretrained("albert-base-v2")

# Tokenize text
tokens_52a = sp_pretrained.tokenize(text_data_5)
token_ids_52a = sp_pretrained.convert_tokens_to_ids(tokens_52a)

# Display tokens and IDs
df_52a = pd.DataFrame({'Token': tokens_52a, 'Token ID': token_ids_52a})
print("=== 5.2 (a) SENTENCEPIECE TOKENIZATION (PRETRAINED ALBERT) ===")
display(df_52a.head(20))
```

5.2 (b) SentencePiece Tokenizer Trained from Scratch

```python
import sentencepiece as spm
import pandas as pd

# 1. Save input text for SentencePiece training
with open("sp_corpus.txt", "w", encoding="utf-8") as f:
    f.write(text_data_5)

# 2. Train SentencePiece model from scratch
spm.SentencePieceTrainer.train(
    input='sp_corpus.txt',
    model_prefix='custom_sp',
    vocab_size=min(500, len(set(text_data_5.split())) + 20),
    model_type='unigram'
)

# 3. Load trained model
sp = spm.SentencePieceProcessor()
sp.load('custom_sp.model')

# 4. Tokenize text
tokens_52b = sp.encode_as_pieces(text_data_5)
token_ids_52b = sp.encode_as_ids(text_data_5)

# Display tokens and IDs
df_52b = pd.DataFrame({'Token': tokens_52b, 'Token ID': token_ids_52b})
print("=== 5.2 (b) SENTENCEPIECE TOKENIZATION (FROM SCRATCH) ===")
display(df_52b.head(20))
```

Experiment 5.3: POS Tagging on Given Sentence
5.3 (a) POS Tagging using spaCy

```python
import spacy
import pandas as pd

nlp = spacy.load("en_core_web_sm")
sentence_53 = "The young student is reading an interesting book in the library."

doc = nlp(sentence_53)

pos_spacy = []
for token in doc:
    pos_spacy.append({
        'Token': token.text,
        'POS Tag': token.pos_,
        'Description': spacy.explain(token.pos_)
    })

df_53a = pd.DataFrame(pos_spacy).drop_duplicates(subset=['Token'])
print("=== 5.3 (a) POS TAGGING (SPACY) ===")
display(df_53a)
```

5.3 (b) POS Tagging using NLTK

```python
import nltk
nltk.download('tagsets_json', quiet=True)
from nltk.tokenize import word_tokenize
import pandas as pd

sentence_53 = "The young student is reading an interesting book in the library."

# Tokenize and Tag
tokens = word_tokenize(sentence_53)
tagged = nltk.pos_tag(tokens)

# NLTK POS Tag descriptions
nltk_pos_descriptions = {
    'DT': 'Determiner',
    'JJ': 'Adjective',
    'NN': 'Noun, singular or mass',
    'VBZ': 'Verb, 3rd person singular present',
    'VBG': 'Verb, gerund or present participle',
    'IN': 'Preposition or subordinating conjunction',
    '.': 'Punctuation mark'
}

pos_nltk = []
for token, tag in tagged:
    desc = nltk_pos_descriptions.get(tag, nltk.help.upenn_tagset(tag))
    pos_nltk.append({
        'Token': token,
        'POS Tag': tag,
        'Description': desc
    })

df_53b = pd.DataFrame(pos_nltk).drop_duplicates(subset=['Token'])
print("=== 5.3 (b) POS TAGGING (NLTK) ===")
display(df_53b)
```

Experiment 5.4: POS Tagging with Frequency Analysis (spaCy)

```python
import spacy
from collections import Counter
import pandas as pd

nlp = spacy.load("en_core_web_sm")

# Process uploaded text from 5.1
doc_54 = nlp(text_data_5)

# Count frequencies of unique tokens (ignoring whitespace/newlines)
valid_tokens = [token.text.lower() for token in doc_54 if not token.is_space]
token_frequencies = Counter(valid_tokens)

# Extract unique token attributes
pos_54_data = []
seen_tokens = set()

for token in doc_54:
    token_lower = token.text.lower()
    if not token.is_space and token_lower not in seen_tokens:
        seen_tokens.add(token_lower)
        pos_54_data.append({
            'Unique Token': token.text,
            'POS Tag': token.pos_,
            'Description': spacy.explain(token.pos_),
            'Frequency': token_frequencies[token_lower]
        })

df_54 = pd.DataFrame(pos_54_data).sort_values(by='Frequency', ascending=False).reset_index(drop=True)

print("=== EXPERIMENT 5.4: POS TAGGING & FREQUENCY ANALYSIS (SPACY) ===")
display(df_54.head(25)) # Displays top 25 unique tokens
```
